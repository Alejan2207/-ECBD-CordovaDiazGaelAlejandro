```python
# Eliminación de registros duplicados
df = df.drop_duplicates()
```

```python
# Revisar valores nulos antes de limpiar
df.isnull().sum()
```

```python
#Transformacion / limpieza de valores faltantes
df["director"] = df["director"].fillna("No especificado")
df["cast"] = df["cast"].fillna("No especificado")
df["country"] = df["country"].fillna("No especificado")
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])
df["date_added"] = df["date_added"].fillna("No especificado")
df["duration"] = df["duration"].fillna("No especificado")
```

```python
# Transformar release_year a tipo entero
df["release_year"] = df["release_year"].astype(int)
```

```python
# Verificar nuevamente valores nulos
df.isnull().sum()
```

```python
# Verificar duplicados después de la limpieza
df.duplicated().sum()
```

```python
#Justificacion
#No fue necesario eliminar duplicados porque el dataset no presentó registros repetidos.
```

```python
#------3. Analisis descriptivo (CAPA DE INFORMACION)
df["type"].value_counts().plot(kind="bar")

plt.title("Cantidad de contenido por tipo")
plt.xlabel("Tipo")
plt.ylabel("Cantidad")
plt.show()
```

```python
df["type"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Distribución de contenido por tipo")
plt.ylabel("")
plt.show()
```

```python
df["release_year"].plot(kind="hist", bins=20)

plt.title("Distribución de años de lanzamiento")
plt.xlabel("Año de lanzamiento")
plt.ylabel("Frecuencia")
plt.show()
```

```python
df["country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 países con más contenido")
plt.xlabel("País")
plt.ylabel("Cantidad")
plt.xticks(rotation=45)
plt.show()
```

```python
#------4. Interpretacion de patrones (CAPA DE CONOCIMIENTO)
#Conclusion 1:
#Netflix cuenta con una mayor cantidad de películas que series, 
#lo que indica que su catálogo está más enfocado en contenido individual que en producciones por temporadas.
```

```python
#Conclusion 2:
#La mayoría del contenido se concentra en años recientes, 
#mostrando que la plataforma ha incrementado su catálogo con producciones modernas y actualizadas.
```

```python
#Conclusion 3: 
#Estados Unidos aparece como uno de los países con mayor cantidad de contenido,
#lo que refleja una fuerte presencia de producciones estadounidenses dentro del dataset analizado.
```
